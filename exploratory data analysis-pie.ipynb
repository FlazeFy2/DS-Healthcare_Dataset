```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("modified_healthcare_dataset.csv")

# Exploratory Data Analysis (EDA) - Pie Chart Most Context for Nonconditional Data
target_col = ['Gender','Blood Type','Medical Condition','Insurance Provider','Admission Type','Test Results']

for dt in target_col:
    # Dataset
    top_context = df[dt].value_counts()
    
    # Pie Chart
    plt.figure(figsize=(8,8))
    top_context.plot.pie(autopct='%1.1f%%', startangle=90)
    plt.title(f"Comparison Of {dt}")
    plt.ylabel("")
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Dataset
df = pd.read_csv("modified_healthcare_dataset.csv")

# Exploratory Data Analysis (EDA) - Pie Chart Patient's Blood Rhesus Comparison
df['Rhesus'] = df['Blood Type'].str[-1]

# Count Rhesus
rhesus_counts = df['Rhesus'].value_counts()

# Pie Chart
plt.figure(figsize=(6, 6))
rhesus_counts.plot.pie(autopct='%1.1f%%', startangle=90, labels=['Positive (+)', 'Negative (-)'])
plt.title("Patient's Blood Rhesus Comparison")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Dataset
df = pd.read_csv("modified_healthcare_dataset.csv")

# Exploratory Data Analysis (EDA) - Pie Chart Patient Distribution by Floor's Room
# Filter valid Room Number
df = df[pd.to_numeric(df['Room Number'], errors='coerce').notnull()]
df['Room Number'] = df['Room Number'].astype(int)
df = df[(df['Room Number'] >= 101) & (df['Room Number'] <= 599)]

# Extract Floor from Room Number
df['Floor'] = df['Room Number'].astype(str).str[0]
df['Floor'] = 'Floor ' + df['Floor']

# Count Floor
floor_counts = df['Floor'].value_counts().sort_index()

# Pie Chart
plt.figure(figsize=(6, 6))
floor_counts.plot.pie(autopct='%1.1f%%', startangle=90)
plt.title("Patient Distribution by Floor's Room")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Dataset
# Exploratory Data Analysis (EDA) - Pie Chart Patient Distribution by Length of Stay
df = pd.read_csv("modified_healthcare_dataset.csv")

# Convert Length of Stay to week buckets
def categorize_stay(days):
    if days < 7:
        return "Under 1 Week"
    elif days < 14:
        return "1–2 Weeks"
    elif days < 21:
        return "2–3 Weeks"
    else:
        return "3+ Weeks"

df["Length of Stay Category"] = df["Length of Stay"].apply(categorize_stay)

# Count the number of patients in each category
stay_counts = df["Length of Stay Category"].value_counts().sort_index()

# Pie Chart
plt.figure(figsize=(6, 6))
stay_counts.plot.pie(autopct='%1.1f%%', startangle=90)
plt.title("Patient Distribution by Length of Stay")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Dataset
# Exploratory Data Analysis (EDA) - Pie Chart Patient Distribution by Admission Quartal
df = pd.read_csv("modified_healthcare_dataset.csv")

# Convert 'Date of Admission' to Quartal buckets
def categorize_quartal(date):
    month = pd.to_datetime(date).month
    if month <= 3:
        return "Quartal - 1"
    elif month <= 6:
        return "Quartal - 2"
    elif month <= 9:
        return "Quartal - 3"
    else:
        return "Quartal - 4"

df["Admission Quartal"] = df["Date of Admission"].apply(categorize_quartal)

# Count the number of patients in each quartal
quartal_counts = df["Admission Quartal"].value_counts().sort_index()

# Pie Chart
plt.figure(figsize=(6, 6))
quartal_counts.plot.pie(autopct='%1.1f%%', startangle=90)
plt.title("Patient Distribution by Admission Quartal")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Dataset
# Exploratory Data Analysis (EDA) - Pie Chart Patient Distribution by Billing Amount Category
df = pd.read_csv("modified_healthcare_dataset.csv")

# Convert 'Billing Amount' to Category Billing
def categorize_billing(amount):
    if amount <= 10000:
        return "Cheap"
    elif amount <= 30000:
        return "Moderate"
    elif amount <= 60000:
        return "Expensive"
    else:
        return "Very Expensive"

df["Billing Amount Category"] = df["Billing Amount"].apply(categorize_billing)

# Count the number of patients in each category
quartal_counts = df["Billing Amount Category"].value_counts().sort_index()

# Pie Chart
plt.figure(figsize=(6, 6))
quartal_counts.plot.pie(autopct='%1.1f%%', startangle=90)
plt.title("Patient Distribution by Billing Amount Category")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Dataset
# Exploratory Data Analysis (EDA) - Pie Chart Patient Distribution by Day Admission (Weekday vs Weekend)
df = pd.read_csv("modified_healthcare_dataset.csv")

# Convert 'Date of Admission' to datetime
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"])

# Create 'Day Type' column based on weekday (0-4) or weekend (5-6)
df["Day Type"] = df["Date of Admission"].apply(lambda x: "Weekend" if x.weekday() >= 5 else "Weekday")

# Count Patients
day_type_counts = df["Day Type"].value_counts().sort_index()

# Pie Chart
plt.figure(figsize=(6, 6))
day_type_counts.plot.pie(autopct='%1.1f%%', startangle=90)
plt.title("Patient Distribution by Day Admission (Weekday vs Weekend)")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Dataset
# Exploratory Data Analysis (EDA) - Pie Chart Patient Distribution by Day - Type Admission
df = pd.read_csv("modified_healthcare_dataset.csv")

# Convert 'Date of Admission' to datetime
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"])

# Create 'Day Type' column based on weekday (0-4) or weekend (5-6)
df["Day Type"] = df["Date of Admission"].apply(lambda x: "Weekend" if x.weekday() >= 5 else "Weekday")

# Create 'Combined Type' column by merging 'Day Type' and 'Admission Type'
df["Combined Type"] = df["Day Type"] + " - " + df["Admission Type"]

# Count Patients by Combined Type
combined_counts = df["Combined Type"].value_counts().sort_index()

# Pie Chart
plt.figure(figsize=(7, 7))
combined_counts.plot.pie(autopct='%1.1f%%', startangle=90)
plt.title("Patient Distribution by Day - Type Admission")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

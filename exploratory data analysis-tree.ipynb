```python
import pandas as pd
import matplotlib.pyplot as plt
import squarify

# Exploratory Data Analysis (EDA) - Tree Map Most Medication Used In Treatment With Total Patient
# Dataset
df = pd.read_csv("modified_healthcare_dataset.csv")

total_context = df.groupby('Medication')['Name'].count().reset_index().head(20)

# Sort 
total_context = total_context.sort_values(by='Name', ascending=False)

# Extract
sizes = total_context['Name']
labels = total_context.apply(lambda x: f"{x['Medication']}\n({x['Name']})", axis=1)

# Plot Tree Map
plt.figure(figsize=(12, 8))
squarify.plot(sizes=sizes, label=labels, alpha=0.8)
plt.title("Most Medication Used In Treatment With Total Patient", fontsize=14)
plt.axis('off')
plt.show()
```
